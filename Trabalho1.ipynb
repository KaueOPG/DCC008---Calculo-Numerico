# Exercício 1



```python
import numpy as np
t = 0.1
n = np.linspace(1,10)
e = n/10 - n*t
```

O código acima inicia-se importando a biblioteca numpy como np, e em seguida declara a variável do tipo float "t" como 0.1 e "n" como um array formado por 50 números com intervalos de passos iguais entre o número 1 e 10, através da função linspace. E por final na saída, é declarado "e", um array em que para cada valor de n em índice x, existe um valor e no mesmo índice, resultante da seguinte fórmula:

$e = \frac{n}{10} - nt$

Como se observa abaixo:

```python
print('n:', n)
print('e:',e)
```

O código acima(importação das bibliotecas numpy e time) foi utilizado para os programas abaixos que seguem um padrão comum, declaram uma variável do tipo float "x" = 1.0, e dentro do while(repetição) fazem o x recerber a metade dele, em seguida printa ao valor dele com a função print e em sleep define um tempo se 0.5 segundos até a execução da proxima linha. Entretanto elas se diferem na condição do while, ou seja, até quando a repetição deve ser executada, sendo para o primeiro quando a soma da variavél mais um tem que ser maior que um, no segundo e no terceiro, equanto a soma da váriavel mais ela mesma seja maior que ela.
```python
# programa 1
x = 1.0
y = 1
while 1+x > 1:
  x = x/2
  print(y, ': ', x)
  y = y+1
  time.sleep(0.5)
```

# Exercício 3



```python

def taylor_sin(x):
  s = 0;
  t = x;
  n = 1;
  while s+t != s:
    s = s + t;
    t = -x**2/((n+1)*(n+2))*t;
    n = n + 2;
  return s
```

A) O laço irá finalizar quando t assumir um valor tão baixo que ao somar com s, por conta do limite mínimo que um número pode ser considerado no código, o t não será considerado na conta, então s mais t será praticamente s mais zero, resultando em s e terminado com o laço.

B)

I- O resultado tem uma precisão próxima de np.sin(), sendo pior em uns e melhor em outros.

III - O uso de séries e os efeitos da aritmética de ponto flutuante na avaliação de funções é válido, demonstrando uma boa precisão.

```python
pi = 3.14
x1 = pi/2
print('taylor_sin(): ', taylor_sin(x1))
print('np.sin(): ', np.sin(x1))
erroa = 1 - taylor_sin(x1)
error = erroa/taylor_sin(x1)
print(error)
erroa = 1 - np.sin(x1)
error = erroa/np.sin(x1)
print(error)
```
