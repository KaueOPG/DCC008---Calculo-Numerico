1 - A matrize esparçs(S) e densa(A) foram recuperadas dos arquivos pelo código a seguir:

```python
A1 =S1.todense('C')
A2 =S2.todense('C')
A3 =S3.todense('C')
A4 =S4.todense('C')
```

2 - Os fatos L(matriz triangular inferior), U(Matriz triangular superior) e P(matrizes ou vetores de permutação) foram obtidos através da função scipy.linalg.lu(A):

```python
P1,L1,U1=scipy.linalg.lu(A1)
P2,L2,U2=scipy.linalg.lu(A2)
P3,L3,U3=scipy.linalg.lu(A3)
P4,L4,U4=scipy.linalg.lu(A4)
```

4 - Para calcular a taxa de preenchimento é necessário utilixar a fórmula dada, sendo que para calcular nnz(A) foi necessário retornar as matrizes densas para sua forma esparça pela função scipy.sparse.csr_matrix(A):

```python
l1 = scipy.sparse.csr_matrix(L1)
u1 = scipy.sparse.csr_matrix(U1)

l2 = scipy.sparse.csr_matrix(L2)
u2 = scipy.sparse.csr_matrix(U2)

l3 = scipy.sparse.csr_matrix(L3)
u3 = scipy.sparse.csr_matrix(U3)

l4 = scipy.sparse.csr_matrix(L4)
u4 = scipy.sparse.csr_matrix(U4)
```

```python
nnza1 = S1.count_nonzero()
nnzl1 = l1.count_nonzero()
nnzu1 = u1.count_nonzero()

nnza2 = S2.count_nonzero()
nnzl2 = l2.count_nonzero()
nnzu2 = u2.count_nonzero()

nnza3 = S3.count_nonzero()
nnzl3 = l3.count_nonzero()
nnzu3 = u3.count_nonzero()

nnza4 = S4.count_nonzero()
nnzl4 = l4.count_nonzero()
nnzu4 = u4.count_nonzero()
```

```python
taxa = [0 , 0, 0, 0]
taxa[0] = 100 - (nnza1/ (nnzu1 + nnzl1)) * 100
taxa[1] = 100 - (nnza2/ (nnzu2 + nnzl2)) * 100
taxa[2] = 100 - (nnza3/ (nnzu3 + nnzl3)) * 100
taxa[3] = 100 - (nnza4/ (nnzu4 + nnzl4)) * 100
taxa
```

5 - A solução do sistema linear denso Ax = b com b = A.dot(u) e u=numpy.ones((n,1)) usando o método x = numpy.linalg.solve(A,b), pode ser vista abaixo:

```python
x_A = [ 0, 0, 0, 0]
n = A1.shape[0]
u= np.ones((n,1))
b = A1.dot(u)
x_A[0] = np.linalg.solve(A1, b)

n = A2.shape[0]
u= np.ones((n,1))
b = A2.dot(u)
x_A[1] = np.linalg.solve(A2, b)

n = A3.shape[0]
u= np.ones((n,1))
b = A3.dot(u)
x_A[2] = np.linalg.solve(A3, b)

n = A4.shape[0]
u= np.ones((n,1))
b = A4.dot(u)
x_A[3] = np.linalg.solve(A4, b)
```

6 - A solução do sistema linear esparso Sx = b com b = S.dot(u) e u=numpy.ones((n, 1)) usando o método x = scipy.sparse.linalg.spsolve(S,b), pode ser vista abaixo:

```python
x_S = [ 0, 0, 0, 0]
n = A1.shape[0]
u= np.ones((n,1))
b = S1.dot(u)
x_S[0] = scipy.sparse.linalg.spsolve(S1, b)

n = A2.shape[0]
u= np.ones((n,1))
b = S2.dot(u)
x_S[1] = scipy.sparse.linalg.spsolve(S2, b)

n = A3.shape[0]
u= np.ones((n,1))
b = S3.dot(u)
x_S[2] = scipy.sparse.linalg.spsolve(S3, b)

n = A4.shape[0]
u= np.ones((n,1))
b = S4.dot(u)
x_S[3] = scipy.sparse.linalg.spsolve(S4, b)
```

7 - A distância relativa entre a solução exata u e a solução aproximada x, pode ser vista abaixo para ambas matrizes:

```python
Erro_A = [ 0, 0, 0, 0]
n = A1.shape[0]
u= np.ones((n,1))
Erro_A[0] = norm(x_A[0]-u, inf)
n = A2.shape[0]
u= np.ones((n,1))
Erro_A[1] = norm(x_A[1]-u, inf)
n = A3.shape[0]
u= np.ones((n,1))
Erro_A[2] = norm(x_A[2]-u, inf)
n = A4.shape[0]
u= np.ones((n,1))
Erro_A[3] = norm(x_A[3]-u, inf)
```

```python
Erro_S = [0, 0, 0, 0]
n = A2.shape[0]
u= np.ones((n,1))
Erro_S[0] = norm(x_S[0]-u, inf)
n = A2.shape[0]
u= np.ones((n,1))
Erro_S[1] = norm(x_S[1]-u, inf)
n = A3.shape[0]
u= np.ones((n,1))
Erro_S[2] = norm(x_S[2]-u, inf)
n = A4.shape[0]
u= np.ones((n,1))
Erro_S[3] = norm(x_S[3]-u, inf)
```

8 - Calculo da distância relativa entre a matriz original e a matriz resultante da decomposição LU:

```python
Erro = [ 0, 0, 0, 0]
M = P1@L1@U1
Erro[0] = norm(A1-M, inf)/norm(A1, inf)
M = P2@L2@U2
Erro[1] = norm(A2-M, inf)/norm(A2, inf)
M = P3@L3@U3
Erro[2] = norm(A3-M, inf)/norm(A3, inf)
M = P4@L4@U4
Erro[3] = norm(A4-M, inf)/norm(A4, inf)
```

9 - Cálculo da norma do resíduo:

```python
norma_A = [ 0, 0, 0, 0]
n = A1.shape[0]
u= np.ones((n,1))
b = S1.dot(u)
norma_A[0] = np.max(abs(A1 @ x_A[0] - b))
n = A2.shape[0]
u= np.ones((n,1))
b = S2.dot(u)
norma_A[1] = np.max(abs(A2 @ x_A[1] - b))
n = A3.shape[0]
u= np.ones((n,1))
b = S3.dot(u)
norma_A[2] = np.max(abs(A3 @ x_A[2]- b))
n = A4.shape[0]
u= np.ones((n,1))
b = S4.dot(u)
norma_A[3] = np.max(abs(A4 @ x_A[3] - b))
```

```python
norma_S = [ 0, 0, 0, 0]
n = A1.shape[0]
u= np.ones((n,1))
b = S1.dot(u)
norma_S[0] = np.max(abs(A1 @ x_S[0] - b))
n = A2.shape[0]
u= np.ones((n,1))
b = S2.dot(u)
norma_S[1] = np.max(abs(A2 @ x_S[1] - b))
n = A3.shape[0]
u= np.ones((n,1))
b = S3.dot(u)
norma_S[2] = np.max(abs(A3 @ x_S[2] - b))
n = A4.shape[0]
u= np.ones((n,1))
b = S4.dot(u)
norma_S[3] = np.max(abs(A4 @ x_S[3] - b))
```

10 - O número de condicionamento da matriz:

```python
K = [ 0, 0, 0, 0]
K[0] = np.linalg.cond(A1)
K[1] = np.linalg.cond(A2)
K[2] = np.linalg.cond(A3)
K[3] = np.linalg.cond(A4)
K
```
