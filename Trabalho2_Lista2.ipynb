```python
#função que verifica se a matriz A é ou não diagonal dominante
def diagonal_dominante(A):
  n = A.shape[0]
  alfa = np.zeros(n)
  for i in range(n):
     alfa[i]= np.sum( abs(A[i,:]))/abs(A[i,i])-1
  M=np.max(alfa)
  if M<1:
    return True, M
  else:
    return False, M
```

```python
#Calcula a matriz de iteração de Gauss-Seidel

def MS(A):
  D = np.diag(np.diag(A))
  U = np.triu(A)-D
  MS = -inv(A-U)@U
  return MS
```
